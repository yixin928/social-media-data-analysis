```python
df3 = pd.read_csv(r"E:/WeiboSpider-master/Weibo_a/整合.csv",encoding='utf-8-sig')

df3 = df3.reset_index(drop=True)

print(df3)
```

```python
#筛选出50字以上的所有微博内容

def filter_comments(content):
    if len(content) >= 150:
        return content
    else:
        return None

# 对comment列应用函数，生成一个新的Series
new_contents = df3['content'].apply(filter_comments)

# 将新的Series添加到原DataFrame中，生成一个新的DataFrame
new_df = pd.concat([df3, new_contents.rename('new_content')], axis=1)

# 删除new_comment列中为None的行
new_df = new_df.dropna(subset=['new_content'])

# 删除原comment列
new_df = new_df.drop(columns=['content'])

# 将new_comment列重命名为comment
new_df = new_df.rename(columns={'new_content': 'content'})

# 保存为新的csv文件
new_df.to_csv('E:/WeiboSpider-master/Weibodata_DataProcessing/Weibo_Chatgpt1.csv', index=False)

print(new_df)
```

```python
df4 = pd.read_csv("E:/WeiboSpider-master/Weibodata_DataProcessing/Weibo_Chatgpt1.csv")

#仅选择需要的列
df4 = df4[['_id', 'created_at', 'content', 'mblogid']]

df4 = df4.rename(columns={'_id': '用户ID', 'created_at': '评论时间', 'content': '用户发表的关于文心一言或ChatGPT的评论性文本', 'mblogid': '用户名'})

# 增加用户所属平台列，全部填充为B站  
df4['用户所属平台'] = '微博' 
# 增加评论对象列，全部填充为ChatGPT  
df4['评论对象'] = 'Chatgpt'

# 将评论日期转换为指定的格式
df4['评论时间'] = pd.to_datetime(df4['评论时间']).dt.strftime('%Y-%m-%d')

# 重排列顺序
df4 = df4[['用户ID', '用户名', '用户所属平台', '评论时间', '评论对象', '用户发表的关于文心一言或ChatGPT的评论性文本']]


# 使用布尔索引过滤数据
df4 = df4.loc[df4['评论时间'] >= '2023-03-14']

# 将某列的英文逗号替换成中文逗号
df4['用户发表的关于文心一言或ChatGPT的评论性文本'] = df4['用户发表的关于文心一言或ChatGPT的评论性文本'].str.replace(',', '，')

print(df4)

df4.to_csv('E:/WeiboSpider-master/Weibodata_DataProcessing/Weibo_Chatgpt2.csv', index=False)
```
