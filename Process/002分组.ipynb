```python

Data_DF = pd.read_csv("../merged_file_2.csv", encoding="utf-8")
print(Data_DF["用户所属平台"].value_counts())   # 查看各平台用户数量
```

```python

def Group(DF):
    # 将表格按照平台类型进行分组
    Grouped_DF = DF.groupby("用户所属平台")
    # 获取分组
    Group1_DF = Grouped_DF.get_group("B站")
    Group1_DF.to_csv("../Data/merged_file_bz.csv", encoding="utf-8", index=False)
    print(Group1_DF.shape)

    Group2_DF = Grouped_DF.get_group("微博")
    Group2_DF.to_csv("../Data/merged_file_wb.csv", encoding="utf-8", index=False)
    print(Group2_DF.shape)

    Group3_DF = Grouped_DF.get_group("知乎")
    Group3_DF.to_csv("../Data/merged_file_zh.csv", encoding="utf-8", index=False)
    print(Group3_DF.shape)

    Group4_DF = Grouped_DF.get_group("百家号")
    Group4_DF.to_csv("../Data/merged_file_bjh.csv", encoding="utf-8", index=False)
    print(Group4_DF.shape)

Group(Data_DF)
```

```python
bjh_DF = pd.read_csv("../Data/merged_file_bjh.csv", encoding="utf-8")
bz_DF = pd.read_csv("../Data/merged_file_bz.csv", encoding="utf-8")
wb_DF = pd.read_csv("../Data/merged_file_wb.csv", encoding="utf-8")
zh_DF = pd.read_csv("../Data/merged_file_zh.csv", encoding="utf-8")

# 将表格按照评论对象进行分组,百家号只抓取了对文心一言的主题文章故不需分组
def Group_dx(DF, DF_name): 
    Grouped_DF = DF.groupby("评论对象")
    # 获取分组
    Group1_DF = Grouped_DF.get_group("ChatGPT")
    Group1_DF.to_csv("../Data/merged_file_"+ DF_name +"_gpt.csv", encoding="utf-8", index=False)
    
    print(Group1_DF.shape)

    Group2_DF = Grouped_DF.get_group("文心一言")
    Group2_DF.to_csv("../Data/merged_file_"+ DF_name +"_wxyy.csv", encoding="utf-8", index=False)
    print(Group2_DF.shape)


Group_dx(bz_DF, DF_name="bz")
Group_dx(wb_DF, DF_name="wb")
Group_dx(zh_DF, DF_name="zh")
```
