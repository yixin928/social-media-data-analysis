```python
import operator

import pandas as pd
from pyecharts.charts import WordCloud

import pyecharts.options as opts
from ast import literal_eval
```

```python

bjh_DF = pd.read_csv("../Data/merged_file_bjh.csv", encoding="utf-8")
bz_DF = pd.read_csv("../Data/merged_file_bz.csv", encoding="utf-8")
wb_DF = pd.read_csv("../Data/merged_file_wb.csv", encoding="utf-8")
zh_DF = pd.read_csv("../Data/merged_file_zh.csv", encoding="utf-8")
```

```python
# 定义绘制词云图函数Draw
def Draw(DF, CommentObject, DF_name):
    # 获取对应评论对象的文本分词结果
    temp = DF["文本分词"].loc[DF["评论对象"]==CommentObject]

    # 初始化单词计数字典
    word_count_dict = {}

    # 统计每个单词的计数
    for item in temp:
        item_list = literal_eval(item)
        if len(item_list)>0:   # 如果关键词不为空
            for word in item_list:
                word_count_dict[word] = word_count_dict.get(word, 0) + 1

    # 按照单词计数从高到低排序
    word_list = list(word_count_dict.items())
    word_list.sort(key=operator.itemgetter(1), reverse=True)
    print(word_list)

    # 初始化WordCloud对象并设置参数
    wordcloud = WordCloud()
    wordcloud.add(series_name='', data_pair=word_list, word_size_range=[20, 100], shape='diamond')

    #生成HTML文件
    path = '../WordCloudDiagram/' + DF_name + "_" + CommentObject +".html"
    wordcloud.render(path)


# 分别绘制不同平台对于文心一言和ChatGPT的认知文本词云图
Draw(zh_DF, CommentObject="文心一言", DF_name="zh")
Draw(zh_DF, CommentObject="ChatGPT", DF_name="zh")
Draw(wb_DF, CommentObject="文心一言", DF_name="wb")
Draw(wb_DF, CommentObject="ChatGPT", DF_name="wb")
Draw(bz_DF, CommentObject="文心一言", DF_name="bz")
Draw(bz_DF, CommentObject="ChatGPT", DF_name="bz")
Draw(bjh_DF, CommentObject="文心一言", DF_name="bjh")
```
